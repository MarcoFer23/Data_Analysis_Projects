```python
"""
-----------------------------
Gambler's Ruin Simulation
-----------------------------

A gambler starts with an initial amount of money.
At each step:
   - with a probability of 50% he wins 1 unit
   - with a probability of 50% he loses 1 unit

 The process stops when:
   - the gambler reaches the target fortune
   - OR the gambler loses everything (ruin)

 This is a classic example of a random walk.

Added a bit of retrogaming nostalgia through chime tunes!
 
"""

import random
import chime

chime.theme('mario')

# Target fortune
target = 10

# Initial capital of the gambler
capital = 5

# Counter for the number of coin tosses
tosses = 0

# Let's use a while loop to make the process go on until the gambler is ruined (0) or reaches the target amount

while capital > 0 and capital < target:

    # Simulate a fair coin toss:
    # 1 = win
    # 0 = loss
    coin = random.randint(0, 1)

    tosses += 1

    # Update the gambler's capital
    if coin == 1:
        capital += 1
    else:
        capital -= 1

    # Print the current situation
    print(f"Launch {tosses}: capital = {capital}")

# Final result
if capital == target:
    chime.success()
    print(f"\nThe gambler reached the target ({target}) "
          f"after {tosses} launches.")
else:
    chime.error()
    print(f"\nThe gambler went broke after {tosses} launches.")
```
