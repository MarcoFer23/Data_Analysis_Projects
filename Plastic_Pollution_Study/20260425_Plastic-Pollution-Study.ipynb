```python
# First things first, let's import libraries / modules to read the dataset(s):

import warnings # Useful to hide warning messages within the output
import pandas as pd
import seaborn as sns
import plotly.express as px
import matplotlib.pyplot as plt

columns_to_use = ["Year", "Annual plastic production between 1950 and 2019"]
global_plastic_production = pd.read_csv("1. Plastic Production/global-plastics-production_our_world_in_data.csv", sep=",", usecols=columns_to_use)
global_plastic_production = global_plastic_production.rename(columns={"Annual plastic production between 1950 and 2019": "Annual Plastic Production"})
global_plastic_production
```

Part Two: Problems Resulting from Improper Garbage Disposal

Mismanaged Plastic Waste per Capita

```python
# Let's read the dataset:

plastic_waste_per_capita = pd.read_csv("2. Mismanaged Plastic Waste Disposal/mismanaged-plastic-waste-per-capita.csv", sep=",")
plastic_waste_per_capita
```

```python
# Here is a list of instructions in Python to get valuable insights about the datasets being looked at: 

"""
plastic_waste_per_capita.describe
plastic_waste_per_capita.info
plastic_waste_per_capita.shape
plastic_waste_per_capita.columns
plastic_waste_per_capita.isnull().sum()
plastic_waste_per_capita.isna().sum()
"""

# We could use .fillna() and inplace=True to make sure "na" values are managed, however the dataset contains null values which should be managed differently.

"""
plastic_waste_per_capita.fillna(value="")
"""

plastic_waste_per_capita
```

```python
# The dataset contains Nan values; we can use the following code to identify columns containing missing values, and count how many missing values there are:

plastic_waste_per_capita.isna().sum()
```

```python
# After finding out the "Code" columns contains missing values, we can use the following code to find out rows containing missing values onòy:

pwpc_missing_values_rows = plastic_waste_per_capita[plastic_waste_per_capita[["Entity","Code","Year", "Mismanaged plastic waste per capita (kg per year)"]].isna().any(axis=1)]
pwpc_missing_values_rows
```

```python
# We can define a mapping to fill missing data within the dataset:

mapping = {
    "Africa": "AF",
    "Asia": "AS",
    "Europe": "EU",
    "North America": "NA",
    "Oceania": "OC"
}
```

```python
# 'Code' column values are missing, we can use the following snippet to assign values based on the 'Entity' column values, using a mapping:

plastic_waste_per_capita["Code"] = plastic_waste_per_capita["Code"].fillna(plastic_waste_per_capita["Entity"].map(mapping))
plastic_waste_per_capita
```

```python
# The following snippet is useful to sort a column's values in descending order:

pwpc_ordered = plastic_waste_per_capita.sort_values(by="Mismanaged plastic waste per capita (kg per year)", ascending=False)
pwpc_ordered
```

```python
# Selecting only the top 10 Countries in terms of "Mismanaged Plastic Waste per Capita":

pwpc_ordered_top_10 = pwpc_ordered.iloc[0:10]
pwpc_ordered_top_10
```

```python
# Displaying data through a bar plot:

x = pwpc_ordered_top_10["Entity"]
y = pwpc_ordered_top_10["Mismanaged plastic waste per capita (kg per year)"]

plt.figure(figsize=(16, 4.6)) 
plt.bar(x, y, color ="#4682B4", width = 0.5)
plt.title("Mismanaged Plastic Waste per Capita, Top 10 Countries", fontfamily="sans-serif", fontsize=12.5)
plt.xlabel("", fontfamily="sans-serif", fontsize=12)
plt.ylabel("KG per Year", fontfamily="sans-serif", fontsize=12)
plt.grid(False)
plt.show()
```

Ocean Plastic Waste by Continent

```python
# Let's read the dataset:

continents = pd.read_csv("3. Plastic in Aquatic Environments/Countries-Continents.csv", sep=",")
continents
```

```python
# The following snippet allows to perform a merge, or join, data between datasets, so that we can perform further calculationd and display data related to 'Continents':

ocean_plastic_waste_per_capita_full = pd.merge(
    ocean_plastic_waste_per_capita,
    continents,
    how="inner",
    on="Entity",
    left_on=None,
    right_on=None,
    left_index=False,
    right_index=False,
    sort=True,
    suffixes=("_x", "_y"),
    copy=True,
    indicator=False,
    validate=None,
)
ocean_plastic_waste_per_capita_full
```

```python
# The following snippet is useful to select only some dataset columns and group data for each of them, after making the sum of related values for each row:

continent_ocean_plastic_waste_sum = ocean_plastic_waste_per_capita_full.loc[:,["Entity", "Mismanaged plastic waste to ocean per capita (kg per year)", "Continent"]]
continents_sum = continent_ocean_plastic_waste_sum.groupby("Continent")["Mismanaged plastic waste to ocean per capita (kg per year)"].sum()
continents_sum
```

```python
# Displaying data through a bar plot:

continents_sum = continents_sum.sort_values(ascending=False)

plt.figure(figsize=(10, 4.6))
plt.bar(continents_sum.index, continents_sum.values, color ="#4682B4", width = 0.5)
plt.title("Ocean Plastic Waste per Capita by Continent (Sum)", fontfamily="sans-serif", fontsize=12.5)
plt.xlabel("", fontfamily="sans-serif", fontsize=12)
plt.ylabel("KG per Capita per Year", fontfamily="sans-serif", fontsize=12)  
plt.show()
```

```python
# The following snippet is useful to select only some dataset columns and group data for each of them, after making the mean of related values for each row:

continent_ocean_plastic_waste_mean = ocean_plastic_waste_per_capita_full.loc[:,["Entity", "Mismanaged plastic waste to ocean per capita (kg per year)", "Continent"]]
continents_mean = continent_ocean_plastic_waste_mean.groupby("Continent")["Mismanaged plastic waste to ocean per capita (kg per year)"].mean()
continents_mean
```

```python
# Displaying data through a bar plot:

continents_mean = continents_mean.sort_values(ascending=False)

plt.figure(figsize=(10, 4.6)) 
plt.bar(continents_mean.index, continents_mean.values, color ="#4682B4", width = 0.5)
plt.title("Ocean Plastic Waste per Capita by Continent (Mean)", fontfamily="sans-serif", fontsize=12.5)
plt.xlabel("", fontfamily="sans-serif", fontsize=12)
plt.ylabel("KG per Capita per Year", fontfamily="sans-serif", fontsize=12)  
plt.show()
```
